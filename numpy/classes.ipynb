## Exercise 3

Load data from ``data/ex3_data.npy`` and filter out rows with ``nan`` values.
Report how many rows are dropped during filtration, globally and how many ``nan`` are in each column.

```python
data = np.load("data/ex3_data.npy")
rows_with_na = np.isnan(data).any(axis = 1)
how_many_na_in_cols = np.isnan(data).sum(axis = 0)
data = data[~rows_with_na, :]
print("Rows with nan values:", rows_with_na.sum())
print("Columns with nan values:", how_many_na_in_cols)
```
